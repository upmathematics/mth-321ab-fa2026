```python
# import functions from packages
from sympy import symbols, Function, Eq, solve, lambdify, diff
```

Suppose you are given an autonomous ODE of the form $\displaystyle \frac{dy}{dt} = f(y)$, where $f(y)$ is some continuous function of $y$ with a parameter.

**Example:** Let $f(y) = y^2 + b$ where $b$ is a parameter, $b \in (-\infty, \infty)$.

### 1. Define 1st-Order ODE using symbolic mathematics in Python.

Now, let's define the given ODE in Python. In the given example, $\displaystyle \frac{dy}{dt} = y^2 + b$.

```python
### CONFIGURABLE ###
# define symbols
y, b = symbols('y, b')

# define the function dydt = f(y) symbolically and numerically
def f(y,b):
    f_y = Function('f')(y) # define f(y) as a function of y
    f_y = y**2 + b         # define f(y) = dy/dt
    return f_y

# print out f(y)
display(f(y,b))
####################
```

### 2. Solve for the critical points and the equilibrium solutions

To solve for the critical points, we set $f(y) = 0$. In the given example, $0 = y^2 + b$.

Now, we use Python to solve for the critical points.

```python
# set up the equation
cp_equation = Eq(f(y,b),0)

# solve for y
cp = solve(cp_equation,y)
```

The results are stored as a list, and the length of this list is equal to the number of solutions.

For this example, there are two critical points that you can print out.

```python
# display the critical points symbolically
for k, i in enumerate(cp):
    print('Critical Point '+str(k+1))
    display(i)
```

So, the equilibrium solutions are $y(t) = -\sqrt{-b}$ and $y(t) = \sqrt{-b}$ for some $b \in (-\infty, \infty)$.

Notice that the equilibrium solutions do not exist when $b > 0$ because that would give a negative value under the square root. Also, If $b = 0$, then there is only one equilibrium solution. We can show this using a phase diagram.

### 3. Determine the stability of equilibriums using linearization.

Through linearization, we take the partial derivative of $f(y)$ with respect to $y$. In the given example, $\displaystyle \frac{\partial f}{\partial y} = 2y$. 
Then, we evaluate $\displaystyle \displaystyle \frac{\partial f}{\partial y}$ at the critical points.

```python
# compute df/dy
derivative = diff(f(y,b),y)

# define the function df/dy symbolically and numerically
def dfdy(y):
    dfdy_y = Function('dfdy')(y) # define df/dy as a function of y
    dfdy_y = derivative          # define df/dy
    return dfdy_y

# print out dfdy
dfdy(y)
```

```python
# display the slopes at the critical points symbolically
for k, i in enumerate(cp):
    print('Critical Point '+str(k+1))
    display(dfdy(y).subs(y,i))
```

Now, we can determine the stability of equilibriums.

* Let $b > 0$.
    * There are two equilibriums $y(t) = -\sqrt{-b}$ and $y(t) = \sqrt{-b}$.
    * Since the $\displaystyle \frac{\partial f}{\partial y} \Bigg|_{y=-\sqrt{-b}} < 0$, then the equilibrium $y(t) = -\sqrt{-b}$ is stable.
    * Since the $\displaystyle \frac{\partial f}{\partial y} \Bigg|_{y=-\sqrt{-b}} > 0$, then the equilibrium $y(t) = \sqrt{-b}$ is unstable.

* Let $b = 0$.
    * The only equilibrium solution is $y(0) = 0$.
    * Since the $\displaystyle \frac{\partial f}{\partial y} \Bigg|_{y=0} = 0$ for $b = 0$, the the equilibrium $y(t) = 0$ is semi-stable (maybe).
    * Since $\displaystyle \frac{\partial^2 f}{\partial y^2} = 2 > 0$ is always positive regardless of the value of $b$, then the solutions around the semi-stable equilibrium is increasing or $\displaystyle \frac{dy}{dt} > 0$

* Let $b < 0$.
    * There are no equilibriums.
 
### 4. Draw the phase diagram and analyze the equilibriums as a parameter changes.

```python
# import functions from packages
from numpy import linspace
from matplotlib.pyplot import subplots, show
from matplotlib.markers import MarkerStyle
from sympy.printing import latex
from ipywidgets import interact, FloatSlider
```

```python
### CONFIGURABLE ###
# define axis limits
y_min, y_max = -5, 5
f_min, f_max = -5, 5

# define parameter limits
b_min, b_max = -5, 5
b_step = 0.25
####################
```

The following code block plots the phase diagram of the given ODE and includes an interactive slider that lets you adjust the parameter in real time.

```python
# convert symbolic functions as numerical functions
f_num = lambdify((y,b),f(y,b),'numpy')
dfdy_num = lambdify((y),dfdy(y),'numpy')

# create a function for solving the critical points
cp_f = Function('f')(b)
cp_f = cp
cp_f_num = lambdify(b,cp_f,'numpy')

# define the plot with parameter b
def plot_f(b=0):

    # generate y and f(y) values
    y_vec = linspace(y_min, y_max, 50) # discretized y
    f_y = f_num(y_vec,b) # f values
    cps = cp_f_num(b) # critical points

    # set-up the figure
    fig, ax = subplots(figsize=(8, 5))

    # draw the x and y axis
    ax.axhline(0, color='black', linewidth=1.5)
    ax.axvline(0, color='black', linewidth=1.5)

    # draw f(y)
    ax.plot(y_vec, f_y, linewidth=2, color='blue', label='$f(y) =$' + f'${latex(f(y,b))}$')

    # draw the slope markers
    for j in range(0,len(y_vec)-1):
        if f_y[j] > 0:
            ax.arrow(y_vec[j], 0, 0.01, 0, head_width=0.2, head_length=0.2, fc='green', ec='green')
        if f_y[j] < 0:
            ax.arrow(y_vec[j], 0, -0.01, 0, head_width=0.2, head_length=0.2, fc='orange', ec='orange')

    # draw the equilibrium points
    for i in cps:
        if dfdy_num(i) < 0:
            ax.plot(i,0, marker='o', linestyle='', markerfacecolor='red', markeredgecolor='red', markersize=12, markeredgewidth=2)
        elif dfdy_num(i) == 0:
            ax.plot(i,0, marker=MarkerStyle('o', fillstyle='left'), linestyle='', markerfacecolor='red', markeredgecolor='red', markersize=12, markeredgewidth=2)
        elif dfdy_num(i) > 0:
            ax.plot(i,0, marker='o', linestyle='', markerfacecolor='None', markeredgecolor='red', markersize=12, markeredgewidth=2)

    # labels
    ax.set_xlim(y_min,y_max)
    ax.set_ylim(f_min,f_max)
    ax.set_xlabel('$y$')
    ax.set_ylabel('$f$')
    ax.set_title('Phase Diagram of $\\frac{dy}{dt} = f(y)$')
    ax.grid(True)
    ax.legend(loc='upper left')
    show()

# plot f(y) and create an interactive slider for parameter b
interact(plot_f, b=FloatSlider(value=0, min=b_min, max=b_max, step=b_step));
```

In the given example, as we change the value of $b$, the phase diagram shows the following:

* The *bifurcation* is when the number of equilibriums changed from two to one, and one to zero, as $b$ changes from negative to positive.
* The *bifurcation point* is at $b=0$ where the number of equilibriums changed as well as its stability.
